**Matplotlib** — это популярная библиотека на языке программирования Python, предназначенная для визуализации данных в виде двумерной (2D) и трехмерной (3D) графики.

Основные возможности

• Создание графиков: строит линейные графики, столбчатые диаграммы, гистограммы, диаграммы рассеяния (scatter), круговые диаграммы и многое другое.

Два подхода к работе

1. Pyplot (matplotlib.pyplot) — процедурный интерфейс (напоминающий синтаксис MATLAB), удобный для быстрого создания простых графиков одной строкой кода.
2. Объектно-ориентированный API (Figure и Axes) — гибкий подход, где график рассматривается как объект холста (Figure) и областей рисования (Axes), что необходимо для сложных и многослойных визуализаций.

```python
import pandas as pd
import numpy as np

# 1. Загрузка данных
try:
    film = pd.read_csv('film.csv') # Загрузка и сохранение данных: чтение и запись файлов в форматах CSV
    inventory = pd.read_csv('inventory.csv')
    rental = pd.read_csv('rental.csv')
    print("✅ Все файлы успешно загружены!")
except FileNotFoundError as e:
    print(f"❌ Ошибка: {e}. Проверьте, что файлы загружены в левую панель (папка 📁).")

# Проверим, что в таблицах есть нужные колонки
print("\nКолонки в film:", film.columns.tolist())
print("Колонки в rental:", rental.columns.tolist())
print("Колонки в inventory:", inventory.columns.tolist())
```

Найти фильмы, которые чаще всего отдают с задержкой и показать их рейтинг

```python


# Убедимся, что колонки с датами имеют правильный формат
#  Python по умолчанию читает даты из CSV просто как текст (строки).
# Функция pd.to_datetime превращает текст в специальный формат "дата и время". Это нужно,
# чтобы потом можно было вычитать одну дату из другой и считать дни.
rental['rental_date'] = pd.to_datetime(rental['rental_date'])
rental['return_date'] = pd.to_datetime(rental['return_date'])

# 1. JOIN
df = pd.merge(film[['film_id', 'title', 'rental_duration']],
              inventory[['film_id', 'inventory_id']],
              on='film_id', how='inner')

# УБИРАЕМ 'rental_id' ИЗ СПИСКА КОЛОНОК
df = pd.merge(df,
              rental[['inventory_id', 'rental_date', 'return_date']],
              on='inventory_id', how='inner')

# 2. WHERE
df = df[df['return_date'].notnull()].copy()

# 3. Вычисление дедлайна и флага просрочки
df['deadline'] = df['rental_date'] + pd.to_timedelta(df['rental_duration'], unit='D')
df['is_overdue'] = np.where(df['return_date'] > df['deadline'], 1, 0)

# 4. GROUP BY
result = df.groupby(['film_id', 'title', 'rental_duration']).agg(
    # ВМЕСТО count('rental_id') ИСПОЛЬЗУЕМ size (считает количество строк)
    total_rentals=('inventory_id', 'size'),
    overdue_rentals=('is_overdue', 'sum')
).reset_index()

# 5. RANK
result['overdue_ratio'] = result['overdue_rentals'] / result['total_rentals']
result['overdue_rank'] = result['overdue_ratio'].rank(method='min', ascending=False).astype(int)

# 6. Сортировка
result = result.sort_values(by='overdue_rank')

print(result[['title', 'total_rentals', 'overdue_rentals', 'overdue_rank']].head(10))
```

```python
# Фильтруем фильмы, у которых есть хотя бы одна просрочка (overdue_rentals > 0)
# И считаем их количество
films_with_overdue = result[result['overdue_rentals'] > 0].shape[0]

# Общее количество фильмов в выборке (для контекста)
total_films_in_result = result.shape[0]

print(f"Всего фильмов в выборке: {total_films_in_result}")
print(f"Из них фильмов с просрочкой (хотя бы 1 раз): {films_with_overdue}")
```

Далее

-- можно посомтреть по каким причинам происходит задержка.

-- например, посмотрим зависимость просрочек от срока аренды (3-7 дней)

```python
# ---------------------------------------------------------
# 2. JOIN (Соединение таблиц)
# ---------------------------------------------------------
# Соединяем film и inventory
df = pd.merge(film[['film_id', 'rental_duration']],
              inventory[['film_id', 'inventory_id']],
              on='film_id', how='inner')

# Соединяем с rental (используем inventory_id для подсчета строк, если нет rental_id)
df = pd.merge(df,
              rental[['inventory_id', 'rental_date', 'return_date']],
              on='inventory_id', how='inner')

# ---------------------------------------------------------
# 3. WHERE (Фильтрация)
# ---------------------------------------------------------
df = df[df['return_date'].notnull()].copy()

# ---------------------------------------------------------
# 4. Вычисление просрочки
# ---------------------------------------------------------
df['deadline'] = df['rental_date'] + pd.to_timedelta(df['rental_duration'], unit='D')

# Флаг просрочки: 1 если просрочено, 0 если вовремя (CASE)
df['is_overdue'] = np.where(df['return_date'] > df['deadline'], 1, 0)

# ---------------------------------------------------------
# 5. GROUP BY (Группировка по rental_duration)
# ---------------------------------------------------------
result = df.groupby('rental_duration').agg(
    total_rentals=('inventory_id', 'size'),  # COUNT(r.rental_id) - считаем количество строк
    overdue_count=('is_overdue', 'sum')      # SUM(CASE WHEN ...)
).reset_index()

# ---------------------------------------------------------
# 6. Расчет процента и переименование колонок
# ---------------------------------------------------------
# ROUND(SUM(...) / COUNT(...) * 100, 1)
result['overdue_percent'] = (result['overdue_count'] / result['total_rentals'] * 100).round(1)

# Переименовываем колонку, как в SQL
result = result.rename(columns={'rental_duration': 'rental_days'})

# ---------------------------------------------------------
# 7. Сортировка (ORDER BY rental_days)
# ---------------------------------------------------------
result = result.sort_values(by='rental_days')

# Вывод результата
print(result.to_string(index=False))
```

вывод:

-- Когда магазин дает всего 3 дня,
-- клиенты видимо не успевают посмотреть фильм, и 72% из них опаздывают.
-- Когда дают 7 дней — опаздывают в 2.5 раза реже (29%).


Далее можно посмотреть: насколько сильно задерживают

```python
# ---------------------------------------------------------
# 2. JOIN и WHERE (Аналог CTE overdue_stats)
# ---------------------------------------------------------
df = pd.merge(film[['film_id', 'title', 'rental_duration']],
              inventory[['film_id', 'inventory_id']],
              on='film_id', how='inner')

df = pd.merge(df,
              rental[['inventory_id', 'rental_date', 'return_date']],
              on='inventory_id', how='inner')

# WHERE r.return_date IS NOT NULL
df = df[df['return_date'].notnull()].copy()

# Вычисляем, на сколько дней опоздали (EXTRACT(DAY FROM ...))
df['deadline'] = df['rental_date'] + pd.to_timedelta(df['rental_duration'], unit='D')
df['days_late'] = (df['return_date'] - df['deadline']).dt.days

# ---------------------------------------------------------
# 3. CASE (Аналог CTE categorized_stats)
# ---------------------------------------------------------
# Задаем условия и соответствующие им категории
conditions = [
    (df['days_late'] <= 0),
    (df['days_late'].between(1, 2)),
    (df['days_late'].between(3, 5)),
    (df['days_late'].between(6, 10)),
    (df['days_late'] > 10)
]
choices = ['Вовремя', '1-2 дня', '3-5 дней', '6-10 дней', 'Более 10 дней']

# Создаем колонку late_category
df['late_category'] = np.select(conditions, choices, default='Неизвестно')

# ---------------------------------------------------------
# 4. GROUP BY и ORDER BY (Финальный SELECT)
# ---------------------------------------------------------
# Группируем по категориям и считаем количество (COUNT(*))
result = df.groupby('late_category').size().reset_index(name='rental_count')

# Для правильной сортировки (ORDER BY CASE ...) зададим порядок категорий
category_order = ['Вовремя', '1-2 дня', '3-5 дней', '6-10 дней', 'Более 10 дней']

# Превращаем колонку в категориальный тип с нужным порядком
result['late_category'] = pd.Categorical(result['late_category'], categories=category_order, ordered=True)

# Сортируем по этой колонке (теперь Pandas знает правильный порядок)
result = result.sort_values('late_category')

# Вывод результата
print(result.to_string(index=False))
```

Выводы:
-- Большинство опозданий — это 1-2 дня (3 482 случая)
-- и 3-5 дней (2 726 случаев)
-- большая часть прокатов возвращается вовремя (9 458 случаев)
-- и незначительная часть с задержкой на 6-10 суток (195 случаев)

```python
# Настройка стиля графиков (чтобы было красиво)
plt.style.use('ggplot')
```

```python
# --- ПОДГОТОВКА ДАННЫХ (из наших предыдущих шагов) ---
film = pd.read_csv('film.csv')
inventory = pd.read_csv('inventory.csv')
rental = pd.read_csv('rental.csv')

rental['rental_date'] = pd.to_datetime(rental['rental_date'])
rental['return_date'] = pd.to_datetime(rental['return_date'])

df = pd.merge(film[['film_id', 'title', 'rental_duration']],
              inventory[['film_id', 'inventory_id']], on='film_id', how='inner')
df = pd.merge(df, rental[['inventory_id', 'rental_date', 'return_date']],
              on='inventory_id', how='inner')
df = df[df['return_date'].notnull()].copy()

df['deadline'] = df['rental_date'] + pd.to_timedelta(df['rental_duration'], unit='D')
df['days_late'] = (df['return_date'] - df['deadline']).dt.days
df['is_overdue'] = np.where(df['days_late'] > 0, 1, 0)
```

```python
# Создаем данные для 1-го графика (по срокам аренды)
duration_stats = df.groupby('rental_duration').agg(
    total_rentals=('inventory_id', 'size'),
    overdue_count=('is_overdue', 'sum')
).reset_index()
duration_stats['overdue_percent'] = (duration_stats['overdue_count'] / duration_stats['total_rentals']) * 100
```

```python
# =====================================================================
# 1. СТОЛБЧАТАЯ ДИАГРАММА (Bar chart)
# =====================================================================
plt.figure(figsize=(10, 5))
plt.bar(duration_stats['rental_duration'].astype(str), duration_stats['overdue_percent'], color='skyblue', edgecolor='black')
plt.title('Процент просрочек в зависимости от срока аренды')
plt.xlabel('Срок аренды (дней)')
plt.ylabel('Процент просрочек (%)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

ВЫВОД: Чем короче срок аренды (3 дня), тем выше процент просрочек (более 70%).
При сроке 6-7 дней клиенты успевают посмотреть фильм вовремя (просрочки падают до ~30%).

```python
# =====================================================================
# 2. ГИСТОГРАММА (Histogram)
# =====================================================================

overdue_days = df[df['days_late'] > 0]['days_late']

plt.figure(figsize=(10, 5))

# КЛЮЧЕВЫЕ ИЗМЕНЕНИЯ:
# 1. bins=6 (так как у нас целые значения от 1 до 6)
# 2. width=1.0 и edgecolor='white' (или None, если нужен сплошной цвет)
plt.hist(overdue_days, bins=15, color='salmon', width=1.0, edgecolor='white')

plt.title('Распределение количества дней просрочки')
plt.xlabel('На сколько дней опоздали')
plt.ylabel('Количество прокатов (частота)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

ВЫВОД: Распределение имеет "правый хвост". Подавляющее большинство клиентов опаздывают всего на 1-5 дней. Случаи задержки более 7 дней встречаются крайне редко.

```python
# Создаем данные для 3-го графика (по фильмам)
film_stats = df.groupby('title').agg(
    total_rentals=('inventory_id', 'size'),
    overdue_count=('is_overdue', 'sum')
).reset_index()
film_stats['overdue_percent'] = (film_stats['overdue_count'] / film_stats['total_rentals']) * 100
```

```python
# =====================================================================
# 3. ТОЧЕЧНАЯ ДИАГРАММА (Scatter plot)
# =====================================================================
# Сравним общее количество прокатов фильма и процент его просрочек
plt.figure(figsize=(10, 5))
plt.scatter(film_stats['total_rentals'], film_stats['overdue_percent'], alpha=0.5, color='purple')
plt.title('Зависимость процента просрочки от популярности фильма')
plt.xlabel('Общее количество прокатов фильма (популярность)')
plt.ylabel('Процент просрочек (%)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

ВЫВОД: Явной линейной зависимости нет. Есть фильмы-лидеры проката (справа),
у которых процент просрочки может быть как низким (20%), так и высоким (80%).
Это значит, что сам по себе фильм не виноват в задержках (сюжет не влияет на возврат), проблема скорее в человеческом факторе или сроках аренды.

```python


# =====================================================================
# 4. ТЕПЛОВАЯ КАРТА (Heatmap)
# =====================================================================
# Для тепловой карты нужны две категориальные переменные.
# Возьмем: 1) Категорию задержки (наши корзины) и 2) Срок аренды (дни).
# Сначала создадим корзины задержки
conditions = [
    (df['days_late'] <= 0),
    (df['days_late'].between(1, 2)),
    (df['days_late'].between(3, 5)),
    (df['days_late'] > 5)
]
choices = ['Вовремя', '1-2 дня', '3-5 дней', 'Более 5 дней']
df['late_category'] = np.select(conditions, choices, default='Неизвестно')

# Считаем, сколько прокатов попало в каждую комбинацию (Срок аренды x Категория задержки)
heatmap_data = pd.crosstab(df['rental_duration'], df['late_category'])

# Рисуем тепловую карту вручную через matplotlib (imshow)
fig, ax = plt.subplots(figsize=(8, 5))
im = ax.imshow(heatmap_data, cmap='YlOrRd', aspect='auto')

# Настраиваем оси
ax.set_xticks(np.arange(len(heatmap_data.columns)))
ax.set_yticks(np.arange(len(heatmap_data.index)))
ax.set_xticklabels(heatmap_data.columns)
ax.set_yticklabels(heatmap_data.index)

# Добавляем подписи
ax.set_xlabel('Категория задержки')
ax.set_ylabel('Срок аренды (дней)')
ax.set_title('Тепловая карта: Срок аренды vs Категория задержки')

# Добавляем цветовую шкалу (colorbar) справа
cbar = ax.figure.colorbar(im, ax=ax)
cbar.ax.set_ylabel("Количество прокатов", rotation=-90, va="bottom")

# Пишем цифры прямо в ячейки
for i in range(len(heatmap_data.index)):
    for j in range(len(heatmap_data.columns)):
        text = ax.text(j, i, heatmap_data.iloc[i, j], ha="center", va="center", color="black")

plt.tight_layout()
plt.show()
```

ВЫВОД:
1. 7-дневный срок аренды работает лучше всего: клиенты успевают посмотреть фильм и меньше всего нарушают правила, 496 задержек фильмов на 1 - 2 дня - могут быть связаны с тем, что клиенты забывают вернуть или не могут вернуть на 7 день (возможно по личным обстоятельствам).
2. Аналогично, для 6 дней срока аренды, с небольшой разницей, есть задержки на 3-5 дней 189 прокатов, но это самая небольшая цифра по задержкам.
3. На коротких сроках (3 дня) самое большое количество задержек 1102 на срок 3-5 дней и 784 на срок 1-2 дня, аналогично, на сроках аренды в 4 дня

Тепловая карта наглядно показывает, что короткие сроки аренды (3-4 дня) — это не слишком эффективно. Увеличение срока до 6-7 дней может улучшить показатель возвратов вовремя.
