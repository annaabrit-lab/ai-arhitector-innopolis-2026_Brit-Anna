1. Импорт библиотек

```python
import random
import statistics
```

2. **Задача 1.** Принимать три параметра (a: int, b: int, length: int) и возвращать список (list) длины length из равномерно
распределенной случайной величины между числами a и b

```python
def generate_random_list(a: int, b: int, length: int) -> list:
    """
    Возвращает список длины length из равномерно распределенной
    случайной величины между числами a и b.
    """
    # random.uniform возвращает число с плавающей точкой в диапазоне [a, b]
    return [random.uniform(a, b) for _ in range(length)]
```

3. **Задача 2.** Принимать созданный список (list) и считать количество уникальных значений переменной (строить частотный
ряд) в виде dict. Например, для списка [1, 2, 3, 1, 2, 2] будет возвращать {1 : 2, 2 : 3, 3 : 1}

```python
def get_frequency_dict(input_list: list) -> dict:
    """
    Принимает список и возвращает словарь с частотным рядом.
    Пример: [1, 2, 2, 3] -> {1: 1, 2: 2, 3: 1}
    """
    freq_dict = {}
    for item in input_list:
        # Если элемент уже есть в словаре, увеличиваем счетчик, иначе создаем
        freq_dict[item] = freq_dict.get(item, 0) + 1
    return freq_dict
```

4. **Задача 3.** Считать минимум, среднюю, медиану, моду, максимум для значений списка

```python
def analyze_statistics(input_list: list):
    """
    Считает минимум, среднее, медиану, моду и максимум для значений списка.
    """
    # На всякий случай сортируем список (хотя statistics.median сама сортирует)
    sorted_list = sorted(input_list)

    # Вычисления
    min_val = min(sorted_list)
    max_val = max(sorted_list)
    mean_val = statistics.mean(sorted_list)
    median_val = statistics.median(sorted_list)

    # Мода может вызвать ошибку, если все значения уникальны (в случайных float это вероятно)
    try:
        mode_val = statistics.mode(sorted_list)
    except statistics.StatisticsError:
        mode_val = "Мода отсутствует (все значения уникальны)"

    return {
        "Минимум": min_val,
        "Среднее": mean_val,
        "Медиана": median_val,
        "Мода": mode_val,
        "Максимум": max_val
    }
```

5. Тестирование

```python
# 1. Генерируем список (например, 10 чисел от 0 до 5)
my_list = generate_random_list(0, 5, 10)
print(f"Сгенерированный список: \n{my_list}\n")

# 2. Строим частотный ряд
freq_dict = get_frequency_dict(my_list)
print(f"Частотный ряд (словарь): \n{freq_dict}\n")

# 3. Считаем статистику
stats = analyze_statistics(my_list)
print("Статистические показатели:")
for key, value in stats.items():
    print(f"{key}: {value}")
```
