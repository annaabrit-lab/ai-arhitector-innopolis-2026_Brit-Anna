**Задание 1.** Создайте одномерный массив из 15 случайных целых чисел от 1 до 100. Выведите его форму (shape), тип данных
(dtype) и количество элементов (size)

```python
# Создаем массив из 15 случайных целых чисел от 1 до 100
arr_1 = np.random.randint(1, 101, size=15)

print("Массив:", arr_1)
print("Форма (shape):", arr_1.shape)
print("Тип данных (dtype):", arr_1.dtype)
print("Количество элементов (size):", arr_1.size)
```

**Коментарии, понятия, определения**

- Кортеж (15,) — краткая справка о том, как данные устроены - размерность массива. Метаданные (информация о данных).
- Форма (15,) означает, что массив является одномерным (вектором).
- np.random.randint (целые числа).
- np.random.uniform - тип данных (dtype) float64 (числа с плавающей точкой).

**Задание 2.** Из массива, созданного в пункте 1, отберите с помощью булевой индексации все значения больше 50. Посчитайте,
сколько таких значений и какую долю они составляют от общего числа элементов

```python
# Создаем булеву маску (условие > 50)
mask = arr_1 > 50

# Применяем маску для отбора значений
filtered_arr = arr_1[mask]

# Считаем количество
count_above_50 = filtered_arr.size # или len(filtered_arr)

# Считаем долю от общего числа элементов
total_elements = arr_1.size
proportion = count_above_50 / total_elements

print("Значения больше 50:", filtered_arr)
print("Количество значений больше 50:", count_above_50)
print(f"Доля от общего числа: {proportion:.2%}") # Вывод в процентах
```

**Коментарии, понятия, определения**

***Булева индексация*** (boolean indexing) — это способ фильтрации данных в библиотеках программирования (например, в NumPy по Руководству индексации или Pandas), когда элементы выбираются с помощью массива логических значений (True и False), который называют булевой маской.

***f-строка*** (от английского formatted string — форматированная строка). Всё, что внутри фигурных скобок {...}, Python воспринимает не как обычный текст, а как код или переменную, и подставляет туда реальное значение.

**Задание 3.** Создайте массив из 10 значений времени ответа оператора поддержки в секундах (случайные целые числа).
Переведите его в минуты одной строкой (без цикла). Вычислите среднее, минимальное и максимальное значение

```python
# Создаем массив из 10 значений (например, от 30 до 600 секунд)
time_seconds = np.random.randint(30, 601, size=10)
print("Время в секундах:", time_seconds)

# Переводим в минуты одной строкой (векторизованная операция)
time_minutes = np.round(time_seconds / 60, 2)
print("Время в минутах:", time_minutes)

# Вычисляем статистику
mean_time = np.mean(time_minutes)
min_time = np.min(time_minutes)
max_time = np.max(time_minutes)

print(f"Среднее значение: {mean_time:.2f} мин.")
print(f"Минимальное значение: {min_time:.2f} мин.")
print(f"Максимальное значение: {max_time:.2f} мин.")
```

**Задание 4.** Создайте матрицу 4×3 (например, 4 студента, 3 оценки по предметам для каждого, целые числа от 1 до 5).
Посчитайте среднюю оценку по каждому клиенту и среднюю оценку по каждому предмету

```python
# Создаем матрицу 4x3 (4 студента, 3 предмета) с оценками от 1 до 5
grades_matrix = np.random.randint(1, 6, size=(4, 3))

print("Матрица оценок (строки - студенты, столбцы - предметы):")
print(grades_matrix)
print()

# Средняя оценка по каждому студенту (по строкам, axis=1)
mean_per_student = np.mean(grades_matrix, axis=1).astype(int)
print("Средняя оценка по каждому студенту:", mean_per_student)

# Средняя оценка по каждому предмету (по столбцам, axis=0)
mean_per_subject = np.mean(grades_matrix, axis=0).astype(int)
print("Средняя оценка по каждому предмету:", mean_per_subject)
```

**Коментарии, понятия, определения**

1. **параметр axis** в функциях np.mean():

axis=1 — считает среднее по строкам.

axis=0 — считает среднее по столбцам.

2. **.astype(int)** берет каждый элемент массива и переводит его в новый тип — целое число.
